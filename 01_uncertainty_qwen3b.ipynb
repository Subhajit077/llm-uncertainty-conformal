```python
import os
os.environ["VLLM_ATTENTION_BACKEND"] = "TRITON_ATTN"  # T4 can't use FlashAttention

from vllm import LLM, SamplingParams
llm = LLM("Qwen/Qwen2.5-3B-Instruct", dtype="float16",
          gpu_memory_utilization=0.90, max_model_len=2048,
          enforce_eager=True)
tok = llm.get_tokenizer()

def make_prompt(q):
    msgs = [{"role":"system","content":"Answer with a short phrase only, no explanation."},
            {"role":"user","content":q}]
    return tok.apply_chat_template(msgs, tokenize=False, add_generation_prompt=True)

out = llm.generate([make_prompt("Who wrote Hamlet?")], SamplingParams(temperature=0, max_tokens=32))
print(out[0].outputs[0].text)
```

```python
import re, string, json, time
from datasets import load_dataset

N = 5000
ds = load_dataset("mandarjoshi/trivia_qa", "rc.nocontext", split="validation").select(range(N))

def norm(s):
    s = s.lower()
    s = "".join(c for c in s if c not in string.punctuation)
    s = re.sub(r"\b(a|an|the)\b", " ", s)
    return " ".join(s.split())

def is_correct(pred, aliases):
    p = norm(pred)
    return any(norm(a) and norm(a) in p for a in aliases)

prompts = [make_prompt(x["question"]) for x in ds]
greedy  = SamplingParams(temperature=0, max_tokens=32, logprobs=0)
sampled = SamplingParams(temperature=1.0, n=10, max_tokens=32)

t0 = time.time()
g_out = llm.generate(prompts, greedy)
print("greedy done:", round(time.time()-t0), "s")
s_out = llm.generate(prompts, sampled)
print("sampled done:", round(time.time()-t0), "s")

records = []
for x, g, s in zip(ds, g_out, s_out):
    aliases = x["answer"]["aliases"]
    go = g.outputs[0]
    g_text = go.text.strip()
    token_lps = [list(d.values())[0].logprob for d in go.logprobs]
    records.append({
        "question": x["question"],
        "aliases": aliases,
        "greedy": g_text,
        "greedy_correct": bool(is_correct(g_text, aliases)),
        "greedy_logprob_sum": sum(token_lps),
        "greedy_logprob_mean": sum(token_lps) / max(len(token_lps), 1),
        "samples": [o.text.strip() for o in s.outputs],
        "samples_correct": [bool(is_correct(o.text.strip(), aliases)) for o in s.outputs],
    })

json.dump(records, open("/kaggle/working/triviaqa_qwen3b.json", "w"))
acc = sum(r["greedy_correct"] for r in records) / len(records)
print("greedy accuracy:", acc)
print(records[0])
```

```python
import json, re, string, random, numpy as np
recs = json.load(open("/kaggle/working/triviaqa_qwen3b.json"))

def norm(s):
    s = s.lower()
    s = "".join(c for c in s if c not in string.punctuation)
    s = re.sub(r"\b(a|an|the)\b", " ", s)
    return " ".join(s.split())

def is_correct(pred, aliases):
    p = " " + norm(pred) + " "
    return any(na and (" " + na + " ") in p for na in map(norm, aliases))

for r in recs:
    r["greedy_correct"] = bool(is_correct(r["greedy"], r["aliases"]))
    r["samples_correct"] = [bool(is_correct(s, r["aliases"])) for s in r["samples"]]

print("accuracy (word-boundary):", np.mean([r["greedy_correct"] for r in recs]))
random.seed(0)
for r in random.sample([r for r in recs if r["greedy_correct"]], 12):
    print(f"CORRECT | {r['question'][:60]} | {r['greedy']} | {r['aliases'][:2]}")
for r in random.sample([r for r in recs if not r["greedy_correct"]], 12):
    print(f"WRONG   | {r['question'][:60]} | {r['greedy']} | {r['aliases'][:2]}")
```

```python
import torch, itertools
from collections import Counter
from transformers import AutoTokenizer, AutoModelForSequenceClassification
from sklearn.metrics import roc_auc_score, brier_score_loss

dev = "cuda:1"
nli_tok = AutoTokenizer.from_pretrained("microsoft/deberta-large-mnli")
nli = AutoModelForSequenceClassification.from_pretrained("microsoft/deberta-large-mnli").to(dev).half().eval()
ENT = [i for i, l in nli.config.id2label.items() if l.lower().startswith("entail")][0]

@torch.no_grad()
def entails(pairs, bs=64):
    out = []
    for i in range(0, len(pairs), bs):
        b = pairs[i:i+bs]
        enc = nli_tok([p for p, h in b], [h for p, h in b], return_tensors="pt",
                      padding=True, truncation=True, max_length=128).to(dev)
        out += (nli(**enc).logits.argmax(-1) == ENT).tolist()
    return out

def semantic_entropy(question, samples):
    uniq = list(dict.fromkeys(norm(s) for s in samples if norm(s)))
    if len(uniq) <= 1:
        return 0.0
    pairs, idx = [], []
    for i, j in itertools.combinations(range(len(uniq)), 2):
        a, b = f"{question} {uniq[i]}", f"{question} {uniq[j]}"
        pairs += [(a, b), (b, a)]; idx.append((i, j))
    res = entails(pairs)
    parent = list(range(len(uniq)))
    def find(x):
        while parent[x] != x: x = parent[x]
        return x
    for k, (i, j) in enumerate(idx):
        if res[2*k] and res[2*k+1]:
            parent[find(i)] = find(j)
    cluster_of = {u: find(i) for i, u in enumerate(uniq)}
    counts = Counter(cluster_of[norm(s)] for s in samples if norm(s))
    p = np.array(list(counts.values()), float); p /= p.sum()
    return float(-(p * np.log(p)).sum())

for n, r in enumerate(recs):
    g = norm(r["greedy"])
    r["selfcons"] = float(np.mean([norm(s) == g for s in r["samples"]]))
    r["sem_ent"] = semantic_entropy(r["question"], r["samples"])
    if n % 500 == 0: print(n, flush=True)

def ece(conf, y, bins=10):
    conf, y = np.asarray(conf), np.asarray(y, float)
    edges = np.linspace(0, 1, bins + 1); e = 0
    for lo, hi in zip(edges[:-1], edges[1:]):
        m = (conf > lo) & (conf <= hi)
        if m.any(): e += m.mean() * abs(y[m].mean() - conf[m].mean())
    return e

y = np.array([r["greedy_correct"] for r in recs])
scores = {
    "token_prob (exp sum logprob)": np.exp([r["greedy_logprob_sum"] for r in recs]),
    "mean token logprob":           np.array([r["greedy_logprob_mean"] for r in recs]),
    "self-consistency":             np.array([r["selfcons"] for r in recs]),
    "semantic entropy (neg)":       -np.array([r["sem_ent"] for r in recs]),
}
print(f"{'method':32s} AUROC   ECE    Brier")
for name, s in scores.items():
    auc = roc_auc_score(y, s)
    if name.startswith(("token_prob", "self-cons")):
        print(f"{name:32s} {auc:.3f}  {ece(s, y):.3f}  {brier_score_loss(y, s):.3f}")
    else:
        print(f"{name:32s} {auc:.3f}    -      -")
json.dump(recs, open("/kaggle/working/triviaqa_qwen3b_scored.json", "w"))
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import cross_val_predict

rng = np.random.default_rng(0)
def boot_ci(s, y, B=1000):
    a = []
    for _ in range(B):
        i = rng.integers(0, len(y), len(y))
        a.append(roc_auc_score(y[i], s[i]))
    return np.percentile(a, [2.5, 97.5])

for name, s in scores.items():
    lo, hi = boot_ci(np.asarray(s), y)
    print(f"{name:32s} AUROC CI [{lo:.3f}, {hi:.3f}]")

X = np.column_stack([
    [r["greedy_logprob_sum"] for r in recs],
    [r["greedy_logprob_mean"] for r in recs],
    [r["selfcons"] for r in recs],
    [r["sem_ent"] for r in recs],
])
clf = make_pipeline(StandardScaler(), LogisticRegression())
p = cross_val_predict(clf, X, y, cv=5, method="predict_proba")[:, 1]
lo, hi = boot_ci(p, y)
print(f"{'combined (5-fold CV)':32s} AUROC {roc_auc_score(y, p):.3f} CI [{lo:.3f}, {hi:.3f}]  "
      f"ECE {ece(p, y):.3f}  Brier {brier_score_loss(y, p):.3f}") 
```

```python
def paired_ci(s1, s2, y, B=1000):
    d = []
    for _ in range(B):
        i = rng.integers(0, len(y), len(y))
        d.append(roc_auc_score(y[i], s1[i]) - roc_auc_score(y[i], s2[i]))
    return np.percentile(d, [2.5, 97.5])

tp = np.asarray(scores["token_prob (exp sum logprob)"])
se = np.asarray(scores["semantic entropy (neg)"])
print("combined - token_prob  diff CI:", paired_ci(p, tp, y).round(4))
print("semantic - token_prob  diff CI:", paired_ci(se, tp, y).round(4))

names = ["logprob_sum", "logprob_mean", "selfcons", "sem_ent"]
print(f"{'single signal, Platt-calibrated':34s} AUROC   ECE    Brier")
for j, n in enumerate(names):
    pj = cross_val_predict(make_pipeline(StandardScaler(), LogisticRegression()),
                           X[:, [j]], y, cv=5, method="predict_proba")[:, 1]
    print(f"{n:34s} {roc_auc_score(y, pj):.3f}  {ece(pj, y):.3f}  {brier_score_loss(y, pj):.3f}")
```

```python
from datasets import load_dataset
mm = load_dataset("cais/mmlu", "all", split="test")
letters = ["A", "B", "C", "D"]
let_ids = [tok.encode(l, add_special_tokens=False)[0] for l in letters]

def mmlu_prompt(x):
    opts = "\n".join(f"{l}. {c}" for l, c in zip(letters, x["choices"]))
    msgs = [{"role": "system", "content": "Answer with the letter of the correct option only."},
            {"role": "user", "content": f"{x['question']}\n{opts}\nAnswer:"}]
    return tok.apply_chat_template(msgs, tokenize=False, add_generation_prompt=True)

outs = llm.generate([mmlu_prompt(x) for x in mm],
                    SamplingParams(temperature=0, max_tokens=1, logprobs=20))

P = np.full((len(mm), 4), -1e9)
for i, o in enumerate(outs):
    lp = o.outputs[0].logprobs[0]
    for k, t in enumerate(let_ids):
        if t in lp: P[i, k] = lp[t].logprob
print("all 4 letters found:", (P > -1e8).all(1).mean())

Z = np.exp(P - P.max(1, keepdims=True)); probs = Z / Z.sum(1, keepdims=True)
labels = np.array(mm["answer"]); subjects = np.array(mm["subject"])
print("MMLU accuracy:", (probs.argmax(1) == labels).mean())
np.savez("/kaggle/working/mmlu_qwen3b.npz", probs=probs, labels=labels, subjects=subjects)
```

```python
def qhat_lac(pr, y, alpha):
    s = np.sort(1 - pr[np.arange(len(y)), y]); n = len(s)
    return s[min(int(np.ceil((n + 1) * (1 - alpha))), n) - 1]

def evaluate(q, pr, y):
    sets = (1 - pr) <= q
    return sets[np.arange(len(y)), y].mean(), sets.sum(1).mean()

rng = np.random.default_rng(0); alpha = 0.1
uniq = np.unique(subjects); iid, shift = [], []
for _ in range(100):
    perm = rng.permutation(len(labels)); h = len(perm) // 2
    a, b = perm[:h], perm[h:]
    iid.append(evaluate(qhat_lac(probs[a], labels[a], alpha), probs[b], labels[b]))
    m = np.isin(subjects, rng.permutation(uniq)[:len(uniq) // 2])
    shift.append(evaluate(qhat_lac(probs[m], labels[m], alpha), probs[~m], labels[~m]))

for name, r in [("random split (iid)", iid), ("subject shift", shift)]:
    r = np.array(r)
    print(f"{name:20s} coverage mean {r[:,0].mean():.3f} std {r[:,0].std():.3f} "
          f"min {r[:,0].min():.3f} | set size {r[:,1].mean():.2f}")

# per-subject coverage under one iid calibration: where does it break?
perm = rng.permutation(len(labels)); h = len(perm) // 2
q = qhat_lac(probs[perm[:h]], labels[perm[:h]], alpha)
rows = []
for s in uniq:
    i = np.where(subjects == s)[0]
    rows.append((evaluate(q, probs[i], labels[i])[0], s, len(i)))
for c, s, n in sorted(rows)[:8]: print(f"{c:.2f}  {s} (n={n})") 
```

```python
acc_s = {s: (probs[subjects == s].argmax(1) == labels[subjects == s]).mean() for s in uniq}
order = sorted(uniq, key=lambda s: -acc_s[s])
easy, hard = order[:len(order)//2], order[len(order)//2:]
me, mh = np.isin(subjects, easy), np.isin(subjects, hard)

# harsh shift: calibrate on easy subjects, test on hard ones
q = qhat_lac(probs[me], labels[me], alpha)
c, s = evaluate(q, probs[mh], labels[mh])
print(f"easy->hard (no fix)      coverage {c:.3f} | set size {s:.2f}")

# Fix A: recalibrate with k labeled target samples
idx_h = np.where(mh)[0]
for k in [50, 100, 200, 500, 1000]:
    cov, sz = [], []
    for _ in range(200):
        p = rng.permutation(idx_h); ci, ti = p[:k], p[k:]
        a, b = evaluate(qhat_lac(probs[ci], labels[ci], alpha), probs[ti], labels[ti])
        cov.append(a); sz.append(b)
    print(f"recalibrate k={k:4d}       coverage {np.mean(cov):.3f} (std {np.std(cov):.3f}) | set size {np.mean(sz):.2f}")

# Fix B: per-subject thresholds vs one global threshold
g_worst, s_worst, g_bad, s_bad = [], [], [], []
for _ in range(30):
    cal_mask = np.zeros(len(labels), bool)
    for s_ in uniq:
        i = rng.permutation(np.where(subjects == s_)[0]); cal_mask[i[:len(i)//2]] = True
    qg = qhat_lac(probs[cal_mask], labels[cal_mask], alpha)
    gc, sc = [], []
    for s_ in uniq:
        i = np.where(subjects == s_)[0]
        ic, it = i[cal_mask[i]], i[~cal_mask[i]]
        gc.append(evaluate(qg, probs[it], labels[it])[0])
        sc.append(evaluate(qhat_lac(probs[ic], labels[ic], alpha), probs[it], labels[it])[0])
    g_worst.append(min(gc)); s_worst.append(min(sc))
    g_bad.append(np.mean(np.array(gc) < 0.85)); s_bad.append(np.mean(np.array(sc) < 0.85))
print(f"global thr:      worst-subject cov {np.mean(g_worst):.3f} | frac subjects <0.85: {np.mean(g_bad):.2f}")
print(f"per-subject thr: worst-subject cov {np.mean(s_worst):.3f} | frac subjects <0.85: {np.mean(s_bad):.2f}")
```
